```python
import os, re, glob
import pandas as pd

DATA_DIR = '/content/drive/MyDrive/대시보드/data'
print(sorted(os.listdir(DATA_DIR)))
```

```python
print(os.path.exists(DATA_DIR))
raw_names = os.listdir(DATA_DIR)
for n in raw_names:
    print(repr(n))
```

```python
import unicodedata

def find_files(data_dir, keyword):
    keyword_nfc = unicodedata.normalize('NFC', keyword)
    matched = []
    for fname in os.listdir(data_dir):
        fname_nfc = unicodedata.normalize('NFC', fname)
        if keyword_nfc in fname_nfc and fname.lower().endswith(('.csv', '.zip')):
            matched.append(os.path.join(data_dir, fname))
    return sorted(matched)

gu_files = find_files(DATA_DIR, '시군구')
print(gu_files)
```

```python
def load_disease_year(path):
    y = int(re.search(r'(20\d{2})', os.path.basename(path)).group(1))
    df = pd.read_excel(path, header=2)
    df.columns = [str(c).strip() for c in df.columns]
    df['진료년도'] = y
    return df
```

```python
def load_year(path, **kwargs):
    y = int(re.search(r'(20\d{2})', os.path.basename(path)).group(1))
    df = pd.read_csv(path, encoding='cp949', compression=None, **kwargs)
    df.columns = [c.strip() for c in df.columns]
    df['진료년도'] = y
    return df

df_gu = pd.concat([load_year(f) for f in gu_files], ignore_index=True)
print(df_gu.shape, df_gu['진료년도'].unique())
```

```python
def quick_check(df, name):
    print(f'=== {name} ===')
    print('shape:', df.shape)
    print(df.dtypes)
    print('연도별 행 수:\n', df['진료년도'].value_counts().sort_index())
    print('결측치:', df.isna().sum().sum(), '| 중복행:', df.duplicated().sum())
    display(df.head(3))
    display(df.describe(include='all').T)

quick_check(df_gu, '시군구별')
```

```python
num_cols = ['환자수','명세서청구건수','입내원일수','보험자부담금(선별포함)','요양급여비용총액(선별포함)']
for c in num_cols:
    if df_gu[c].dtype == 'object':
        df_gu[c] = df_gu[c].astype(str).str.strip().astype('int64')
```

```python
region_sets = {y: set(zip(g['시도'], g['시군구'])) for y, g in df_gu.groupby('진료년도')}
years = sorted(region_sets)
for i in range(1, len(years)):
    prev, cur = years[i-1], years[i]
    added = region_sets[cur] - region_sets[prev]
    removed = region_sets[prev] - region_sets[cur]
    if added or removed:
        print(f'{prev} → {cur}')
        print('  추가:', added)
        print('  삭제:', removed)
```

```python
mask_bc = df_gu['시군구'].isin(['부천소사구','부천원미구','부천오정구'])
bc_merged = (df_gu[mask_bc].groupby(['진료년도','시도'], as_index=False)[num_cols].sum())
bc_merged['시군구'] = '부천시'
df_gu = pd.concat([df_gu[~mask_bc], bc_merged], ignore_index=True)

mask_gw = df_gu['시군구'].isin(['군위군', '대구군위군'])
gw_merged = (df_gu[mask_gw]
             .assign(시도='대구', 시군구='대구군위군')
             .groupby(['진료년도', '시도', '시군구'], as_index=False)[num_cols].sum())
df_gu = pd.concat([df_gu[~mask_gw], gw_merged], ignore_index=True)
```

```python
region_sets_after = {y: set(zip(g['시도'], g['시군구'])) for y, g in df_gu.groupby('진료년도')}
years = sorted(region_sets_after)
for i in range(1, len(years)):
    prev, cur = years[i-1], years[i]
    diff = (region_sets_after[cur] - region_sets_after[prev]) | (region_sets_after[prev] - region_sets_after[cur])
    print(prev, '→', cur, ':', diff if diff else '변화 없음')
```

```python
def value_sanity_check(df, cost_total_col, insurer_col):
    print('음수 값 있는 컬럼:', [c for c in num_cols if (df[c] < 0).any()])
    print('0인데 나머지 값은 0이 아닌 행 수:', ((df['환자수']==0) & (df[cost_total_col]>0)).sum())
    # 보험자부담금은 항상 총진료비 이하여야 함
    bad = df[df[insurer_col] > df[cost_total_col]]
    print('보험자부담금 > 총진료비인 이상행:', len(bad))
    if len(bad):
        display(bad.head())

value_sanity_check(df_gu, '요양급여비용총액(선별포함)', '보험자부담금(선별포함)')
```

```python
subject_files = find_files(DATA_DIR, '표시과목')
df_subj = pd.concat([load_year(f) for f in subject_files], ignore_index=True)

quick_check(df_subj, '표시과목별')
```

```python
subj_num_cols = ['환자수','명세서청구건수','입내원일수','보험자부담금(선별포함)','요양급여비용총액(선별포함)']  # 실제 컬럼명은 quick_check 출력 보고 맞춰주세요

for c in subj_num_cols:
    if df_subj[c].dtype == 'object':
        df_subj[c] = df_subj[c].astype(str).str.strip().astype('int64')
```

```python
sido_sets = {y: set(g['시도']) for y, g in df_subj.groupby('진료년도')}
years = sorted(sido_sets)
for i in range(1, len(years)):
    prev, cur = years[i-1], years[i]
    diff = (sido_sets[cur] - sido_sets[prev]) | (sido_sets[prev] - sido_sets[cur])
    print(prev, '→', cur, ':', diff if diff else '변화 없음')
```

```python
value_sanity_check(df_subj, '요양급여비용총액(선별포함)', '보험자부담금(선별포함)')
```

```python
def find_files(data_dir, keyword, exts=('.csv', '.zip')):
    keyword_nfc = unicodedata.normalize('NFC', keyword)
    matched = []
    for fname in os.listdir(data_dir):
        fname_nfc = unicodedata.normalize('NFC', fname)
        if keyword_nfc in fname_nfc and fname.lower().endswith(exts):
            matched.append(os.path.join(data_dir, fname))
    return sorted(matched)
```

```python
# 쉼표 포함 문자열 -> 숫자 변환
for col in ['요양급여비용총액', '보험자부담금']:
    df_subj[col] = pd.to_numeric(
        df_subj[col].astype(str).str.replace(',', ''), errors='coerce'
    )

# 이름이 다른 동일 항목 병합 (2021년 명칭 -> 2022년 이후 명칭으로 통일)
df_subj['명세서청구건수'] = df_subj['명세서청구건수'].combine_first(df_subj['명세서 청구건수'])
df_subj['보험자부담금(선별포함)'] = df_subj['보험자부담금(선별포함)'].combine_first(df_subj['보험자부담금'])
df_subj['요양급여비용총액(선별포함)'] = df_subj['요양급여비용총액(선별포함)'].combine_first(df_subj['요양급여비용총액'])

df_subj = df_subj.drop(columns=['명세서 청구건수', '보험자부담금', '요양급여비용총액'])

quick_check(df_subj, '표시과목별 (병합 후)')
value_sanity_check(df_subj, '요양급여비용총액(선별포함)', '보험자부담금(선별포함)')
```

```python
disease_files = find_files(DATA_DIR, '다빈도', exts=('.xlsx', '.xls'))
print(disease_files)  # 5개 파일 맞는지, 순서 확인

df_disease = pd.concat([load_disease_year(f) for f in disease_files], ignore_index=True)

quick_check(df_disease, '다빈도상병별')
```

```python
# 구분 값이 입원/외래 두 가지인지
print('구분 값:', df_disease['구분'].unique())

# 음수 값 체크
num_cols_disease = ['진료인원', '인당진료비', '인당입(내)원일수']
print('음수 값 있는 컬럼:', [c for c in num_cols_disease if (df_disease[c] < 0).any()])

# 같은 연도+구분 내 순위(1~10) 중복 여부
dup_rank = df_disease.duplicated(subset=['진료년도', '구분', '순위'])
print('순위 중복 행 수:', dup_rank.sum())

# 같은 연도+구분 내 코드 중복 여부
dup_code = df_disease.duplicated(subset=['진료년도', '구분', '코드'])
print('코드 중복 행 수:', dup_code.sum())

# 연도별 순위 1~10이 실제로 다 채워져 있는지
print(df_disease.groupby(['진료년도', '구분'])['순위'].apply(lambda s: sorted(s.tolist())))
```
